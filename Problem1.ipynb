```python
# Student information using Dictionary, Tuple and List

students = {
    101: ("Ved", "Computer Science", 85),
    102: ("Robert", "Mechanical", 78),
    103: ("John", "Electronics", 92)
}

# Add a new student record
students[104] = ("Julie", "Information Technology", 88)

# Delete an existing student record
del students[102]

# Update details of a student
students[103] = ("John", "Electronics", 95)

# Display final student records
print("Final Student Records:")

for roll_no, details in students.items():
    print("Roll Number:", roll_no)
    print("Name:", details[0])
    print("Branch:", details[1])
    print("Marks:", details[2])
    print()
```
